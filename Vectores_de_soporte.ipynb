## Vectores de Soporte (SVM): Un Resumen Completo

Las Máquinas de Vectores de Soporte (SVM) son modelos versátiles de aprendizaje automático utilizados para clasificaciones lineales y no lineales, regresión y detección de anomalías. Son especialmente efectivas con conjuntos de datos pequeños o medianos.

### 1. Clasificación Lineal

Las SVM buscan separar dos clases de datos creando un 'margen' lo más amplio posible entre ellas. Este margen puede ser:
*   **Margen Duro**: Muy estricto, no permite violaciones. Se rompe con datos atípicos.
*   **Margen Blando**: Más flexible, permite algunas violaciones dentro del margen para manejar datos atípicos y ruidosos. Su flexibilidad se controla con el hiperparámetro `C`: un `C` bajo permite más violaciones (margen más ancho), mientras que un `C` alto las penaliza fuertemente (margen más estrecho).

### 2. Clasificación No Lineal

Cuando los datos no pueden separarse linealmente, las SVM proyectan los datos a un espacio de mayor dimensión donde sí pueden ser separados por un hiperplano. Esto permite crear fronteras de decisión curvas y complejas en el espacio original.

### 3. El Truco del Kernel

Para evitar el alto costo computacional de calcular explícitamente las nuevas dimensiones en el espacio transformado, las SVM utilizan el 'Truco del Kernel'. Este truco permite obtener el mismo resultado que si se hubieran proyectado y calculado los puntos en el espacio de mayor dimensión, pero solo utilizando funciones de similitud entre los puntos originales. Kernels comunes incluyen el lineal, polinomial y RBF (Radial Basis Function).

### 4. Similitud y Kernel RBF Gaussiano

El Kernel de Función de Base Radial Gaussiana (RBF) es muy popular. Mide la similitud entre puntos, transformando datos en dimensiones donde pueden ser separados linealmente. El hiperparámetro `gamma` controla el 'ancho' de esta función:
*   **`gamma` alto**: La frontera de decisión se vuelve irregular, ajustándose a los puntos de entrenamiento (posible sobreajuste).
*   **`gamma` bajo**: La frontera es más suave y generalizada (posible infraajuste).

### 5. Regresión SVM (SVR)

A diferencia de la clasificación, la SVR invierte la lógica: en lugar de un margen para separar, busca ajustar la mayor cantidad de puntos dentro de un margen. El objetivo es encontrar una función que se ajuste a los datos con un cierto nivel de tolerancia `ε` (epsilon), minimizando el error fuera de este margen. El parámetro `C` también controla la penalización por errores fuera del margen de tolerancia.

### 6. Matemáticas Internas

El algoritmo de SVM busca minimizar el tamaño del vector de pesos (`w`) para lograr el margen más ancho posible. Esto se logra formulando el problema de optimización de forma primal y luego transformándolo a su **forma dual**. La forma dual es crucial porque permite que los cálculos se realicen eficientemente utilizando únicamente productos escalares, lo cual es fundamental para la aplicación del Truco del Kernel y la escalabilidad del modelo.
