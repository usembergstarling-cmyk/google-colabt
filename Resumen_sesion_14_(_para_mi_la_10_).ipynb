## Aprendizaje por Conjuntos (Ensemble Learning)

### 1. Introducción a los Conjuntos
*   Enseñar a las máquinas cómo varios "aprendices débiles" pueden crear un "aprendiz fuerte".
*   Ejemplo: Una moneda sesgada (51% cara); lanzarla mil veces aumenta la probabilidad de obtener más caras al 75%.

### 2. Clasificadores de Votación
*   Múltiples clasificadores distintos trabajando sobre los mismos datos.
*   Cada modelo emite un voto individual.
*   **Votación dura:** Gana la clase con la mayoría de votos.
*   **Votación blanda:** Promedia las probabilidades de clase de cada predictor.

### 3. Bagging, Pasting y OOB
*   **Bagging (Bootstrap Aggregating):** Muestreo aleatorio de datos de entrenamiento CON reemplazo de instancias. Un mismo dato puede ser escogido varias veces por un predictor.
*   **Pasting:** Muestreo aleatorio de datos de entrenamiento SIN reemplazo. Disminuye la varianza.
*   Los cálculos se reducen solo a segundos.
*   **OOB (Out-Of-Bag):** Las instancias se eligen al azar; el algoritmo procesa solo el 63%, el otro 37% son llamadas instancias "fuera de bolsa" y se usan para validación.

### 4. Bosques Aleatorios (Random Forests)
*   Estrategias de muestreo:
    *   **Parches aleatorios:** Muestrea instancias y características.
    *   **Subespacios aleatorios:** Muestrea solo las características.
    *   **Árboles Extra (Extremely Randomized Trees):** Umbrales de división aleatorios.

### 5. Boosting (Procesamiento Secuencial)
*   **Pasos del Boosting:**
    1.  Entrenar un clasificador inicial.
    2.  Aumentar el peso de las instancias mal clasificadas.
    3.  Entrenar un nuevo clasificador con los pesos ajustados.
    4.  Repetir el ciclo secuencialmente.
*   La corrección secuencial ajusta los errores progresivamente.
*   **Gradient Boosting:** Busca adivinar y corregir el error que dejó el modelo anterior.
*   **Histogram-based Gradient Boosting:** Versión más rápida y eficiente de Gradient Boosting. Agrupa (discretiza) los valores numéricos de las variables en intervalos (histogramas) para procesar grandes cantidades de información.

### 6. Stacking (Aprendizaje Apilado)
*   Se entrena un algoritmo "supremo" llamado mezclador (meta-learner).
*   Los modelos de las primeras capas mandan sus predicciones hacia arriba.
*   El mezclador lanza la predicción final.

## Reducción de Dimensionalidad

### 1. Maldición de la Dimensionalidad
*   Ejemplo: ¿Distancia de 408.25 en un hipercubo de un millón de ejes?
*   En dimensiones altas hay mucho espacio vacío.
*   Alto riesgo de sobreajuste debido a datos extremadamente dispersos en espacios de alta dimensión.

### 2. Proyección y Aprendizaje de Variedades
*   **Proyección:** Reduce la dimensionalidad (ej., datos 3D a 2D). Útil cuando los datos se encuentran en un subespacio plano.
*   **Aprendizaje de Variedades (Manifold Learning):** Para cuando los datos están en una variedad no lineal en un espacio de alta dimensión (ej., una "superficie" enrollada).

### 3. Tutorial Técnico: PCA (Análisis de Componentes Principales)
*   Encuentra el hiperplano más cercano a los datos y los proyecta, preservando la máxima varianza.
*   **Pasos:**
    1.  **Centrar:** Centrar las instancias alrededor del origen.
    2.  **Aplicar SVD (Singular Value Decomposition):** Descomponer matrices para hallar los componentes principales (nuevos ejes).
    3.  **Proyectar:** Multiplicar los datos originales por los nuevos ejes d-dimensionales.
*   PCA preserva la mayor varianza y minimiza la distancia cuadrática media.
*   Ejemplo: Reducir un modelo a 154 dimensiones para preservar el 95% de la varianza inicial.

### 4. Optimizando PCA a Escala
*   **Método Incremental:** Memoria baja y trabaja por mini-lotes.
*   **Método Aleatorio:** Más rápido, aprovecha la aleatoriedad.
*   La proyección es eficiente para dimensiones masivas.

### 5. LLE y Otras Técnicas
*   **LLE (Locally Linear Embedding):** Se usa para "desenrollar" estructuras complejas como el "Swiss Roll", preservando las distancias locales.
*   **Otras técnicas:**
    *   **MDS (Multidimensional Scaling):** Busca preservar distancias globales al comprimir el espacio.
    *   **Isomap:** Método de reducción de dimensionalidad no lineal que simplifica datos complejos de alta dimensión mientras conserva su estructura geométrica intrínseca.
    *   **t-SNE (t-Distributed Stochastic Neighbor Embedding):** Algoritmo de aprendizaje automático para reducir la cantidad de dimensiones en conjuntos de datos complejos y mostrarlos en un gráfico de dos o tres dimensiones, ideal para visualización.

## Orden del Caos (Aprendizaje No Supervisado)

*   Información caótica sin organización.
*   Los datos son como océanos sin explorar.

### 1. Datos No Supervisados
*   **Agrupamiento (Clustering):** Identificar y asignar datos similares a grupos basados en sus características.
*   Es el motor detrás de la toma de decisiones; ayuda con la segmentación de clientes en un proyecto. Transforma datos crudos en acciones concretas.

### 2. Agrupamiento con K-Means
*   Identifica grupos distintos y ajusta el centroide de cada grupo.
*   Métricas críticas utilizadas:
    *   **Inercia:** Suma de las distancias cuadráticas de cada punto a su centroide asignado.
    *   **Puntuación de silueta:** Mide qué tan similar es un objeto a su propio clúster en comparación con otros clústeres.

### 3. Segmentación con K-Means
*   **Segmentación de imágenes por color:** Aplicando K-Means, agrupa los colores en clusters, disminuyendo la complejidad visual.
*   **Proceso Iterativo para Mejora:**
    1.  **Entrenar:** Detectar instancias más dudosas.
    2.  **Etiquetar:** Un experto humano revisa y etiqueta los datos.
    3.  **Iterar:** Repetir el proceso hasta optimizar el rendimiento.

### 4. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
*   Algoritmo basado en densidad para identificar grupos continuos de formas arbitrarias.
*   Eficaz para detectar ruido y clústeres de formas irregulares.

### 5. Detección de Anomalías y Novedades
*   **Detección de Anomalías:** Busca aislar eventos disruptivos (datos que se desvían significativamente de la mayoría).
*   **Detección de Novedades:** Busca detectar instancias raras o previamente no vistas en un conjunto de datos (datos anómalos en un conjunto predominantemente "normal").

### 6. Mezclas Gaussianas (Gaussian Mixture Models - GMM)
*   Crea zonas de densidad elípticas que se adaptan a la distribución de los puntos.
*   **Criterios para seleccionar el número óptimo de grupos:**
    *   **BIC (Bayesian Information Criterion)**
    *   **AIC (Akaike Information Criterion)**
    *   Métricas para calcular el número óptimo de grupos en modelos probabilísticos.
