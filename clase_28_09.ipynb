## Despertar de la Memoria: Redes Neuronales Recurrentes

Este resumen aborda conceptos clave sobre redes neuronales recurrentes y su aplicación en el procesamiento de secuencias.

### 1. Redes Recurrentes y Estado Oculto (RNNs)
*   **Definición**: Redes neuronales con conexiones de retroalimentación que mantienen un estado interno ('estado oculto', `h`) a lo largo del tiempo, permitiendo procesar secuencias de cualquier longitud.
*   **Funcionamiento**: Una neurona procesa la entrada actual y su propia salida del paso anterior, comenzando sin un estado pasado inicial.

### 2. Retropropagación Temporal
*   **Proceso**: Método para entrenar RNNs que extiende la retropropagación en el tiempo. Implica un pase hacia adelante (evaluación de la salida), comparación con errores, retropropagación de estos errores y ajuste de pesos (descenso).

### 3. Pronóstico de Series Temporales
*   **Aplicación**: Usado para predecir futuros valores en datos secuenciales (ej. transporte público).
*   **Tipos**: `Univariante` (una variable) o `Multivariante` (múltiples variables como autobuses, trenes, día).
*   **Desafío**: Los errores se acumulan en secuencias largas; los expertos construyen modelos "secuencia a secuencia".

### 4. Limitaciones de Memoria a Corto Plazo
*   Las RNNs simples tienden a "olvidar" información relevante en secuencias que superan los diez pasos temporales.

### 5. Celdas de Memoria Avanzadas
*   **LSTM (Long Short-Term Memory)**: Divide la información en dos "carriles" (corto y largo plazo) y utiliza tres "compuertas lógicas" (olvido, entrada, salida) para controlar el flujo de información y evitar la pérdida de memoria a largo plazo.
*   **GRU (Gated Recurrent Unit)**: Una alternativa más simple y eficiente a LSTM, que combina los carriles y reduce las compuertas, consumiendo menos recursos.

### 6. Procesamiento de Secuencias Largas
*   **Preprocesamiento**: Uso de capas convolucionales unidimensionales para manejar secuencias muy largas sin abrumar a la red.
*   **WaveNet**: Arquitectura que utiliza capas convolucionales dilatadas para procesar miles de pasos de tiempo en paralelo.
    *   **Tasa de Dilatación**: Permite a las capas analizar datos a diferentes escalas temporales; las capas bajas revisan detalles finos, mientras que las capas altas entienden patrones a largo plazo.
