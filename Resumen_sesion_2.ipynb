### 1. Inteligencia Artificial y Machine Learning

La Inteligencia Artificial (IA) busca crear sistemas capaces de aprender de los datos, reconocer patrones, hacer predicciones y tomar decisiones. Dentro de la IA se encuentra el **Machine Learning**, que permite a los modelos aprender a partir de datos para resolver distintos problemas.

### 2. Truco de Kernel

El **truco de Kernel** se utiliza principalmente en algoritmos como las **Máquinas de Soporte Vectorial (SVM)**. Permite resolver problemas que no son separables linealmente al transformar los datos mediante funciones llamadas **kernels**, facilitando su clasificación.

### 3. Evaluación de modelos de clasificación

La **matriz de confusión** permite evaluar un modelo mediante cuatro resultados:

* **Verdaderos positivos (TP)**
* **Verdaderos negativos (TN)**
* **Falsos positivos (FP)**
* **Falsos negativos (FN)**

A partir de estos valores se calculan métricas importantes:

* **Recall:** mide qué proporción de los positivos reales fueron detectados.
* **Precision:** indica qué proporción de las predicciones positivas fueron realmente positivas.
* **F1-score:** combina Precision y Recall para medir su equilibrio.

### 4. Curva ROC y ROC-AUC

La **curva ROC** muestra la relación entre la tasa de verdaderos positivos y la tasa de falsos positivos al modificar el umbral de clasificación.

El **ROC-AUC** representa el área bajo esta curva y permite evaluar qué tan bien el modelo distingue entre las diferentes clases.

### 5. Scikit-learn

**Scikit-learn** es una de las principales bibliotecas de Python para Machine Learning. Permite:

* Preparar datos.
* Entrenar modelos.
* Realizar predicciones.
* Evaluar resultados.
* Optimizar hiperparámetros.

Entre sus algoritmos supervisados se encuentran:

**Regresión Logística, Árboles de Decisión, Random Forest, SVM, KNN y Boosting.**

### 6. Aprendizaje supervisado

El **aprendizaje supervisado** utiliza datos de entrada acompañados de una respuesta conocida o **etiqueta**.

Se divide principalmente en:

* **Clasificación:** predice categorías.
* **Regresión:** predice valores numéricos.

### 7. Bagging y Boosting

Son técnicas que combinan varios modelos para mejorar el rendimiento.

**Bagging:** entrena varios modelos y combina sus resultados para reducir principalmente la variabilidad. Un ejemplo es **Random Forest**.

**Boosting:** entrena modelos de forma secuencial, haciendo que cada nuevo modelo intente corregir los errores de los anteriores. Algunos ejemplos son **AdaBoost, Gradient Boosting y XGBoost**.

### 8. RandomizedSearchCV

**RandomizedSearchCV** es una herramienta de Scikit-learn para encontrar buenos **hiperparámetros**.

En lugar de probar todas las combinaciones posibles, selecciona aleatoriamente un número determinado de combinaciones y las evalúa mediante **validación cruzada**, haciendo la búsqueda más eficiente.

### 9. Idea principal

Desarrollar un buen modelo de Machine Learning no consiste únicamente en entrenarlo. Es necesario:

**Seleccionar el algoritmo adecuado → ajustar los hiperparámetros → entrenar el modelo → evaluar su rendimiento → mejorar su configuración.**

Para evaluar correctamente se pueden utilizar herramientas y métricas como la **matriz de confusión, Precision, Recall, F1-score y ROC-AUC**.
