: Machine Learning
El Machine Learning (ML) permite que un programa mejore su rendimiento en una tarea específica a medida que adquiere experiencia. Para esto, utiliza técnicas como la minería de datos.

1. Tipos de aprendizaje
Supervisado: el algoritmo aprende con datos previamente etiquetados, como si tuviera un “maestro”.
No supervisado: el algoritmo busca patrones y relaciones en los datos sin recibir respuestas correctas.
Por refuerzo: un agente toma decisiones mediante prueba y error, recibiendo recompensas o penalizaciones.
2. Flujo de datos
Aprendizaje por lotes: procesa los datos de una sola vez; puede requerir mucho poder computacional.
Aprendizaje en línea: procesa los datos continuamente y permite adaptarse a información nueva.
Fuera del núcleo: permite trabajar con conjuntos de datos demasiado grandes para caber en la memoria RAM.
3. Estrategias de generalización
Basado en instancias: guarda los datos de entrenamiento y los utiliza cuando recibe una nueva consulta.
Basado en modelos: analiza los datos para crear un modelo o función matemática que represente los patrones generales.
4. Principales desafíos
La calidad de un modelo depende mucho de la limpieza y relevancia de los datos.

Sobreajuste (overfitting): el modelo memoriza los datos y el ruido, por lo que funciona mal con datos nuevos.
Infraajuste (underfitting): el modelo es demasiado simple y no logra aprender correctamente los patrones.
5. Pruebas y validación
Entrenamiento: se enseña al modelo utilizando los datos disponibles.
Validación: se ajustan sus características y parámetros.
Prueba final: se evalúa con datos que el modelo no ha visto.
