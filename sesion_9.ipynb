 Herramientas para Categorización y Clasificación

Cuando trabajamos con variables objetivo discretas (categorías), utilizamos librerías de software diseñadas para automatizar y optimizar estos modelos sin necesidad de programar las operaciones matemáticas desde cero.

---

## 1. Implementación en Scikit-Learn (Ecosistema Clásico de Python)

`scikit-learn` es la librería estándar de facto para Machine Learning tradicional en Python. Ofrece implementaciones optimizadas y listas para usar.

*   **Regresión Logística (Clasificación Binaria y Multiclase):**
    *   **Clase:** `LogisticRegression` del módulo `sklearn.linear_model`.
    *   **Soporte Multiclase:** Detecta automáticamente si el problema es multiclase y aplica por defecto la estrategia *One-vs-Rest* (OVR) o *Multinomial* (Softmax) según el resolvedor (`solver`) elegido.
    *   **Parámetros clave:** `penalty` (para aplicar regularización L1/L2), `C` (fuerza de regularización inversa) y `solver` (algoritmos de optimización como `lbfgs`, `liblinear`, `saga`).

*   **Otros Clasificadores Populares en Scikit-Learn:**
    *   `KNeighborsClassifier`: Clasificación basada en vecinos más cercanos.
    *   `RandomForestClassifier`: Ensamble de árboles de decisión, altamente robusto.
    *   `SVC` (Support Vector Machines): Clasificadores basados en margen máximo.

## 2. Implementación en Deep Learning (Keras y TensorFlow)

Para problemas complejos o de alta dimensionalidad (imágenes, texto largo), utilizamos frameworks de aprendizaje profundo.

*   **Capa de Salida y Activación en Software:**
    *   **Para Clasificación Binaria:** Se utiliza una única neurona en la capa final con la activación `'sigmoid'`. En el entrenamiento se configura la pérdida como `'binary_crossentropy'`.
    *   **Para Clasificación Multiclase:** Se configura la capa final con tantas neuronas como clases tengamos, usando la activación `'softmax'`. La pérdida se define como `'categorical_crossentropy'` (si las etiquetas están en formato de codificación *one-hot*) o `'sparse_categorical_crossentropy'` (si las etiquetas son números enteros).

## 3. Herramientas de Evaluación e Interpretación en Software

`scikit-learn` proporciona un módulo extremadamente completo llamado `metrics` para analizar el rendimiento de nuestros modelos:

*   **Reporte de Clasificación (`classification_report`):** Genera automáticamente una tabla de texto con la Precisión, el Recall (Sensibilidad), el F1-Score y el soporte para cada clase.
*   **Matriz de Confusión Visual (`ConfusionMatrixDisplay`):** Permite pintar y exportar un mapa de calor interactivo que muestra dónde se está confundiendo el modelo.
*   **Curvas ROC y AUC (`RocCurveDisplay`):** Herramientas gráficas para evaluar la capacidad de discriminación del clasificador binario frente a distintos umbrales de decisión.
