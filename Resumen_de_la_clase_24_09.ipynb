## Mejorando el Entrenamiento de Redes Profundas

El entrenamiento de redes neuronales profundas presenta desafíos como el **desvanecimiento y explosión de gradientes** y la **saturación de funciones de activación** (ej. sigmoide). Para mitigar estos problemas, se han desarrollado diversas técnicas:

*   **Inicialización de pesos:** Métodos como Glorot, He y LeCun optimizan la escala de los pesos iniciales, adaptándose a diferentes funciones de activación (tanh, ReLU, ELU, SELU, GELU, Swish, Mish) para asegurar una propagación de señales claras.

*   **Funciones de activación modernas:** Activaciones como ELU y SELU promueven la auto-normalización, manteniendo una media de 0 y desviación estándar de 1, lo que ayuda a estabilizar los gradientes. Otras funciones incluyen GELU, Swish y Mish.

*   **Normalización y estabilización:** La **Normalización por Lotes (Batch Normalization)** centra, escala y desplaza las activaciones de las capas para estabilizar el entrenamiento. Para estructuras neuronales complejas, el **recorte de gradientes (gradient clipping)** limita las actualizaciones de parámetros.

*   **Aprendizaje por transferencia:** Esta técnica revolucionaria permite reutilizar capas de redes profundas pre-entrenadas para nuevas tareas. Consiste en congelar las capas base y ajustar las capas superiores para la tarea específica, reduciendo significativamente los recursos computacionales y el tiempo de entrenamiento.

*   **Pre-entrenamiento y regularización:** El pre-entrenamiento permite a la red aprender detectores de características de alto nivel. La **regularización (ej. Dropout)** desactiva neuronas aleatoriamente durante el entrenamiento, evitando la co-dependencia y mejorando la generalización del modelo, lo que puede reducir la tasa de error global en un 40%.
